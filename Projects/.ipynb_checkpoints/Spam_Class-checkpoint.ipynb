# Data preprocessing

```python
nltk.download('stopwords')
```

```python
import string
```

```python
def text_transfor(text):
    text = text.lower()
    text = nltk.word_tokenize(text)
    y=[]
    for i in text:
        if i.isalnum():
            y.append(i)

    text = y[:]
    y.clear()

    for i in text:
        if i not in stopwords.words('english') and i not in string.punctuation:
            y.append(i)

    text = y[:]
    y.clear()

    for i in text:
        y.append(ps.stem(i))
        
    text = y[:]
    y.clear()
    
    return text
```

```python
text_transfor("My loved name is Chiranjeet && is the")
```

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
df['transformed_text']=df['text'].apply(text_transfor)
```

# Model Building

```python
df.head()
```

```python
df.drop(columns='transformed_text', inplace=True)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
cv = CountVectorizer()
tf = TfidfVectorizer(max_features=3000)
```

```python
X1 = tf.fit_transform(df['transformed_text1']).toarray()
```

```python
X1
```

```python
X = cv.fit_transform(df['transformed_text1']).toarray()
```

```python
y = df['target'].values
```

```python
X,y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report, precision_score,recall_score,f1_score
```

```python
X_train,X_test,y_train,y_test = train_test_split(X1,y,test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
```

```python
gnb = GaussianNB()
mnb = MultinomialNB()
bnb = BernoulliNB()
```

```python
gnb.fit(X_train,y_train)
y_pred = gnb.predict(X_test)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(precision_score(y_test,y_pred))
```

```python
mnb.fit(X_train,y_train)
y_pred1 = mnb.predict(X_test)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred1))
print(confusion_matrix(y_test,y_pred1))
print(classification_report(y_test,y_pred1))
print(precision_score(y_test,y_pred1))
```

```python
bnb.fit(X_train,y_train)
y_pred2 = bnb.predict(X_test)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred2))
print(confusion_matrix(y_test,y_pred2))
print(classification_report(y_test,y_pred2))
print(precision_score(y_test,y_pred2))
```

```python
# tfidf --> MultinomialNB
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
svc = SVC(kernel='sigmoid', gamma=1.0)
knc = KNeighborsClassifier()
mnb = MultinomialNB()
dtc = DecisionTreeClassifier(max_depth=5)
lrc = LogisticRegression(solver='liblinear', penalty='l1')
rfc = RandomForestClassifier(n_estimators=50, random_state=2)
abc = AdaBoostClassifier(n_estimators=50, random_state=2)
bc = BaggingClassifier(n_estimators=50, random_state=2)
etc = ExtraTreesClassifier(n_estimators=50, random_state=2)
gbdt = GradientBoostingClassifier(n_estimators=50,random_state=2)
xgb = XGBClassifier(n_estimators=50,random_state=2)
```

```python
clfs = {
    'SVC' : svc,
    'KN' : knc, 
    'NB': mnb, 
    'DT': dtc, 
    'LR': lrc, 
    'RF': rfc, 
    'AdaBoost': abc, 
    'BgC': bc, 
    'ETC': etc,
    'GBDT':gbdt,
    'xgb':xgb
}
```

```python
def train_classifier(clf,X_train,y_train,X_test,y_test):
    clf.fit(X_train,y_train)
    y_pred = clf.predict(X_test)
    accuracy = accuracy_score(y_test,y_pred)
    precision = precision_score(y_test,y_pred)

    return accuracy,precision
```

```python
train_classifier(svc,X_train,y_train,X_test,y_test)
```

```python
acc_score=[]
prec_score=[]

for name, clf in clfs.items():
    curr_accuracy,curr_prec = train_classifier(clf,X_train,y_train,X_test,y_test)
    
    print("For ",name)
    print("Accuracy - ",curr_accuracy)
    print("Precision - ",curr_prec)
    
    acc_score.append(curr_accuracy)
    prec_score.append(curr_prec)
    
```

```python
dm = pd.DataFrame({'Algorithms':clfs.keys(), 'Accuracy':acc_score, 'Precision':prec_score}).sort_values('Precision',ascending=False)
```

```python
dm
```

```python
temp_df = pd.DataFrame({'Algorithms':clfs.keys(),'Accuracy_max_ft_3000':acc_score,'Precision_max_ft_3000':prec_score}).sort_values('Precision_max_ft_3000',ascending=False)
```

```python
temp_df
```

```python
new_df = dm.merge(temp_df, on="Algorithms")
```

```python
gbdt = GradientBoostingClassifier(n_estimators=50,random_state=2)
mnb = MultinomialNB()
etc = ExtraTreesClassifier(n_estimators=50,random_state=42)

from sklearn.ensemble import VotingClassifier
```

```python
voting = VotingClassifier(estimators=[
    ('gbdt', gbdt),
    ('mnb', mnb), 
    ('etc', etc)
], voting='soft')
```

```python
voting.fit(X_train,y_train)
```

```python
y_pred = voting.predict(X_test)
print("Accuracy",accuracy_score(y_test,y_pred))
print("Precision",precision_score(y_test,y_pred))
```

```python
estimators = [('gbdt', gbdt),('mnb', mnb),('etc', etc)]
final_estimator=RandomForestClassifier()
```

```python
from sklearn.ensemble import StackingClassifier
```

```python
clf = StackingClassifier(estimators=estimators, final_estimator=final_estimator)
```

```python
clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
print("Accuracy",accuracy_score(y_test,y_pred))
print("Precision",precision_score(y_test,y_pred))
```
